# DATA VISUALIZATION

# 1 Bar Plot — Top 10 players by rating

```python
import matplotlib.pyplot as plt

top10 = df.nlargest(10, "rating")

top10.plot(x="full_name", y="rating", kind="bar")
plt.show()
```

# Insights for Bar Plot — Top 10 NBA Players by Rating
- The bar plot compares the ratings of the top 10 NBA players in the dataset.

- The player with the tallest bar has the highest rating among these 10 players.

- Differences in bar heights show the rating gap between players.

- Players with similar bar heights have similar ratings.

- This visualization helps identify the highest-rated players quickly.

# 2 Line Plot — Player rating trend

```python
df["rating"].head(20).plot(kind="line", marker="o")

plt.title("Player Rating Trend")
plt.xlabel("Player Index")
plt.ylabel("Rating")
plt.show()
```

# Insights for Line Plot — NBA Player Rating Trend
- The line plot shows the ratings of the first 20 NBA players in the dataset.

- The line moves upward when the next player's rating is higher and downward when it is lower.

- Peaks represent players with relatively higher ratings, while dips represent lower ratings.

- The graph helps visualize variations in player ratings.



# 3 Histogram — Rating distribution

```python
df["rating"].plot(kind="hist", bins=10)

plt.title("Rating Distribution")
plt.xlabel("Rating")
plt.show()
```

# Insights for Histogram — NBA Player Rating Distribution
- The histogram shows the distribution of ratings among NBA players.

- The tallest bar represents the rating ranges containing the largest number of players.

- Shorter bars indicate rating ranges withs fewer players.

- The histogram helps identify where most players ratings are concentrated.

- The overall shape shows whether ratings are spread out of clustered within a particular range.
    

# 4 Scatter Plot — Player Height vs Weight

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

df.plot(x="height", y="weight", kind="scatter")

plt.title("Player Height vs Weight")
plt.xlabel("Height")
plt.ylabel("Weight")
plt.show()
```

# Insights for Scatter Plot — NBA Player Height vs Weight
- The scatter plot shows the relationship between NBA players' height and weight.

- Each dot represents one player in the dataset.

- Taller players generally tend to have higher body weights.

- Most points may cluster around common height and weight ranges, showing the typical physical profiles of players.

- Points far from the main cluster may represent unusually tall, short, light, or heavy players.
